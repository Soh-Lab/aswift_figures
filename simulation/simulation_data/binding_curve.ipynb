```python
import numpy as np
```

```python
concentrations = np.logspace(-8.0, -4.0, 13)
Kd = 1e-6
frac_bound = concentrations / (Kd + concentrations)
signal_on =  1 + 0.5*(frac_bound)
signal_off = 1 - 0.5*(frac_bound)
```

```python
print(", ".join(f"{v:.3f}" for v in signal_on))

print(", ".join(f"{v:.3f}" for v in signal_off))
```
