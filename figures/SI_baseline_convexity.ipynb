# Doxorubicin and kanamycin baseline-shape diagnostics

Full datasets: 3,000 doxorubicin + 412 kanamycin traces. ATP is excluded. Endpoint-line results describe selected baseline geometry, not true background curvature. Raw-flank slopes may contain peak tails and cannot resolve that limitation.

```python
from pathlib import Path
import pandas as pd
root = Path.cwd() if (Path.cwd()/"revisions").exists() else Path.cwd().parent
endpoints = pd.read_csv(root/"revisions/results/curvature_full/endpoint_summary.csv")
display(endpoints[endpoints.tolerance_fraction == 0.01])
display(endpoints.pivot(index="dataset", columns="tolerance_fraction", values="percent_stays_below_with_tolerance"))
display(endpoints.pivot(index="dataset", columns="tolerance_fraction", values="percent_dips_below"))
```

## Experimental baseline tail-to-peak ratios
Stress-comparable ratio: maximum endpoint baseline rise above the baseline at the fitted peak, divided by peak height. Also report the maximum rise anywhere outside the peak window and absolute (offset-dependent) ratios. Scope: all 3,412 dox/kana traces. These describe estimated backgrounds; shape differences prevent interpreting a ratio as an accuracy guarantee.

```python
tails = pd.read_csv(root/"revisions/results/curvature_full/tail_ratio_summary.csv")
display(tails[tails.metric.str.endswith("rise_over_peak")])
display(tails[tails.metric.str.endswith("absolute_over_peak")])
```

## Corrected smoothed-tail ratio
The requested metric uses smoothed measured current at the tails, minus the selected baseline beneath the peak, divided by fitted peak height.
